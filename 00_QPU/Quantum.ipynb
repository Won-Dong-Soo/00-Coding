5. superposition

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# 1. 큐비트 1개, 클래식 비트 1개
qc = QuantumCircuit(1)

# 2. H 게이트로 중첩 만들기: |0> -> (|0> + |1>)/sqrt(2)
qc.h(0)

# 3. 모든 큐비트 측정
qc = qc.measure_all(inplace=False)

print("회로:")
print(qc)

# 4. Sampler로 실행 (로컬 statevector 시뮬레이터)
sampler = StatevectorSampler()

# shots=1000 번 돌림 (코인 1000번 던지는 느낌)
job = sampler.run([qc], shots=1000)
result = job.result()

# 결과 카운트 꺼내기
counts = result[0].data.meas.get_counts()
print("측정 결과:", counts)

# 5. 히스토그램 플롯
plot_histogram(counts)
plt.show()
```

8. GHZ/GHSH

```python
from qiskit import QuantumCircuit, transpile
from qiskit_ibm_runtime import QiskitRuntimeService, SamplerV2 as Sampler
import numpy as np

# backend 입력
backend_list = []
inp = 'ibm_qasm_simulator'

while inp != "end":
    inp = input("사용할 backend 입력(끝낼 시 end) : ")
    if inp != "end":
        backend_list.append(inp)

# service는 한 번만 생성
service = QiskitRuntimeService()

# CHSH expectation
def expectation(counts):
    total = sum(counts.values())
    return (
        counts.get("00", 0)
        + counts.get("11", 0)
        - counts.get("01", 0)
        - counts.get("10", 0)
    ) / total

# CHSH circuit (핵심 수정 반영)
def chsh_circuit(theta_a, theta_b):
    qc = QuantumCircuit(2)
    qc.h(0)
    qc.cx(0, 1)

    qc.ry(-theta_a, 0)
    qc.ry(-theta_b, 1)

    qc.measure_all()
    return qc

# GHZ circuit
def ghz_circuit():
    qc = QuantumCircuit(3)
    qc.h(0)
    qc.cx(0, 1)
    qc.cx(0, 2)
    qc.measure_all()
    return qc


for backend_name in backend_list:

    backend = service.backend(backend_name)
    print("사용 백엔드:", backend.name, "| qubits:", backend.num_qubits)

    sampler = Sampler(backend)

    # ================= GHZ =================
    ghz_qc = transpile(ghz_circuit(), backend=backend)

    job = sampler.run([ghz_qc], shots=5000)
    result = job.result()
    ghz_counts = result[0].data.meas.get_counts()

    print("\n[GHZ 결과]")
    print(ghz_counts)

    # ================= CHSH =================
    settings = [
        (0, 0),
        (0, np.pi/4),
        (np.pi/2, 0),
        (np.pi/2, np.pi/4)
    ]

    E = []

    for a, b in settings:
        qc = transpile(chsh_circuit(a, b), backend=backend)

        job = sampler.run([qc], shots=5000)
        result = job.result()

        counts = result[0].data.meas.get_counts()
        E.append(expectation(counts))

    S = E[0] + E[1] + E[2] - E[3]

    print("\n[CHSH 결과]")
    print("E values:", E)
    print("CHSH S =", S)
```
