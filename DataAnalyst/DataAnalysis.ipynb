### 서울 기온 분석

```python
import pandas as pd

df = pd.read_csv('seoul_temp.csv', encoding = 'cp949')

print(df)
avg_temp= list(df['평균기온(℃)'])
avg_temp[:10]

date = list(df['날짜'])

date[:10]

for i in range(len(date)):
    date[i] = date[i][1::]

date

type(avg_temp[0])
```

### 지난 10년동안의 매달 평균기온 그래프 출력

```python
_bef10years_data = {
    'month' : [],
    'temp' : []
}
stidx = 0
for i in range(len(date)):
    if date[i][0:4] == '1907':
        stidx = i
        break

monthsum = 0
monthcount = 0
monthidx = int(date[stidx][5:7])
bef_monthidx = 0
past_year = 0
for i in range(stidx, len(date)):
    if date[i] == '':
        continue
    bef_monthidx = monthidx
    monthidx = int(date[i][5:7]) + past_year*12
    if bef_monthidx != monthidx:
        if bef_monthidx%12 == 0 and monthidx%12 == 1:
            past_year += 1
            monthidx += 12
        _bef10years_data['month'].append(bef_monthidx)
        _bef10years_data['temp'].append(monthsum/monthcount)
        monthsum = 0
        monthcount = 0
    monthcount += 1
    monthsum += avg_temp[i]

# _bef10years_data
plt.plot(_bef10years_data['month'], _bef10years_data['temp'])
plt.xlabel('month')
plt.ylabel('avg_temp')
plt.ylim(-30, 50)
plt.show()
```

```python
_bef10years_data = {
    'year' : [],
    'temp' : []
}
stidx = 0
for i in range(len(date)):
    if date[i][0:4] == '1907':
        stidx = i
        break

yearsum = 0
yearcount = 0
yearidx = int(date[stidx][0:4])
bef_yearidx = 0
for i in range(stidx, len(date)):
    if date[i] == '':
        continue
    if date[i][5:7] != "08":
        continue
    bef_yearidx = yearidx
    yearidx = int(date[i][0:4])
    if bef_yearidx != yearidx and yearcount != 0:
        _bef10years_data['year'].append(bef_yearidx)
        _bef10years_data['temp'].append(yearsum/yearcount)
        yearsum = 0
        yearcount = 0
    yearcount += 1
    yearsum += avg_temp[i]

# _bef10years_data
plt.plot(_bef10years_data['year'], _bef10years_data['temp'])
plt.title('August AVG temp')
plt.xlabel('year')
plt.ylabel('avg_temp')
plt.ylim(15, 35)
plt.grid()
plt.show()
```
