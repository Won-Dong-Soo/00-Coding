### 지난 10년동안의 매달 평균기온 그래프 출력

```python
_bef10years_data = {
    'month' : [],
    'temp' : []
}
stidx = 0
for i in range(len(date)):
    if date[i][0:4] == '2014':
        stidx = i
        break

monthsum = 0
monthcount = 0
monthidx = 1
bef_monthidx = 0
past_year = 0
for i in range(stidx, len(date)):
    if date[i] == '':
        continue
    bef_monthidx = monthidx
    monthidx = int(date[i][5:7]) + past_year*12
    if bef_monthidx != monthidx:
        if bef_monthidx%12 == 0 and monthidx%12 == 1:
            past_year += 1
            monthidx += 12
        _bef10years_data['month'].append(bef_monthidx)
        _bef10years_data['temp'].append(monthsum/monthcount)
        monthsum = 0
        monthcount = 0
    monthcount += 1
    monthsum += min_temp[i]

# _bef10years_data
plt.plot(_bef10years_data['month'], _bef10years_data['temp'])
plt.xlabel('month')
plt.ylabel('avg_temp')
plt.show()
```
