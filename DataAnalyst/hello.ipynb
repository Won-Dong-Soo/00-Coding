### 서울 기온 분석

```python
import pandas as pd

df = pd.read_csv('seoul_temp.csv', encoding = 'cp949')

df
min_temp= list(df['최저기온(℃)'])
min_temp[:10]

date = list(df['날짜'])

date[:10]

for i in range(len(date)):
    date[i] = date[i][1::]

date

type(min_temp[0])
```

#### 결측치 확인

```python
cnt = 0
for i in min_temp:
    if i != i:
        cnt += 1
(cnt)
```

```python
null_date = []
for i in range(len(date)):
    if min_temp[i] != min_temp[i]:
        null_date.append(date[i])


null_date, len(null_date)


```

#### 최저기온

```python
_2025_temp = []
for i in range(len(date)):
    if date[i][0:4] == '2025':
        _2025_temp.append(min_temp[i])

mintemp = min(_2025_temp)
mindate = ''
for i in range(len(min_temp)):
    if min_temp[i] == mintemp:
        mindate = date[i]

mintemp, mindate
```
