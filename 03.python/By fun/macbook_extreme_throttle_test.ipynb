## § Monitoring (CPU temp / freq via powermetrics)

```python

import subprocess
import threading

def monitor():
    cmd = ["sudo", "powermetrics", "--samplers", "cpu_power", "-i", "1000", "-S"]
    subprocess.run(cmd)

t = threading.Thread(target=monitor, daemon=True)
t.start()

print("Monitoring started (sudo password may be required)")
```

## § CPU Infinite Burn

```python
import multiprocessing as mp

def burn(_):
    x = 0
    while True:
        x += 1
        x %= 10**9

if __name__ == "__main__":
    mp.set_start_method("fork")  # 핵심

    cores = mp.cpu_count()
    print("Cores:", cores)

    with mp.Pool(cores) as p:
        p.map(burn, range(cores))
```

## § Memory Pressure + Swap

```python

import numpy as np

arrays = []
try:
    while True:
        arrays.append(np.ones((8000, 8000)))
        print(f"Allocated: {len(arrays)} chunks")
except MemoryError:
    print("Memory maxed → swap likely active")
```

## § Disk Saturation

```python

import os

while True:
    with open("stress.bin", "wb") as f:
        f.write(os.urandom(1024 * 1024 * 1024))  # 1GB write
    print("1GB written")
```

## § GPU Max (Metal via PyTorch)

```python

import torch

if torch.backends.mps.is_available():
    device = torch.device("mps")
    x = torch.rand((12000, 12000), device=device)
    while True:
        x = x @ x
        print("GPU step")
else:
    print("MPS not available")
```
