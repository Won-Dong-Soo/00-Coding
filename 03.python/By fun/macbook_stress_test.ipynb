## § CPU Full Load

```python

import multiprocessing as mp
import time

def burn_cpu(x):
    s = 0
    for i in range(10**7):
        s += (i*i) % 1234567
    return s

if __name__ == "__main__":
    cores = mp.cpu_count()
    print(f"Using {cores} cores")
    with mp.Pool(cores) as p:
        start = time.time()
        p.map(burn_cpu, range(cores))
        print("Time:", time.time() - start)
```

## § Memory Stress

```python

import numpy as np

try:
    arrays = []
    for i in range(20):
        print(f"Allocating chunk {i}")
        arrays.append(np.ones((5000, 5000)))
except MemoryError:
    print("Memory limit reached")
```

## § Disk I/O Stress

```python

import os

data = os.urandom(1024 * 1024 * 500)  # 500MB

with open("test.bin", "wb") as f:
    f.write(data)

with open("test.bin", "rb") as f:
    _ = f.read()

print("Disk I/O complete")
```

## § GPU (Metal, Optional - requires PyTorch)

```python

import torch

if torch.backends.mps.is_available():
    device = torch.device("mps")
    print("Using GPU:", device)

    x = torch.rand((10000, 10000), device=device)
    for i in range(10):
        x = x @ x
    print("GPU stress done")
else:
    print("MPS not available")
```
